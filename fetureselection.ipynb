```python
import pandas as pd

data={
    "assignment": [2,4,6,8,10],
    "project":[5,6,6,7,8],
    "marks":[40,50,60,85,95]
}
df =pd.DataFrame(data)
df["internals"] =df["assignment"] + df["project"]
print(df)
df["final_marks"] = df["internals"] + df["marks"]
```

```python
 from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_regression

X= df[["marks","project","internals"]]
y = df[["assignment"]]

selector = SelectKBest(score_func=f_regression, k=2)
X_selected = selector.fit_transform(X, y)
print(X_selected)
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

data={
    "cloud":[2,4,5,6,7],
    "web-pro":[3,6,7,8,8],
    "ml":[4,8,6,9,5]
}
df = pd.DataFrame(data)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
pca = PCA(n_components=2)

principal_components = pca.fit_transform(scaled_data)
print(principal_components)
```

```python
import matplotlib.pyplot as plt

plt.scatter(principal_components[:,0], principal_components[:,1])
plt.xlabel("principal_components 5")
plt.ylabel("principal_components 5")

plt.show()
```
